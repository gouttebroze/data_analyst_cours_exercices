```python
import matplotlib.pyplot as plt
import matplotlib
import numpy as np
matplotlib.__version__
```

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([10, 15, 35, 15])

plt.figure(figsize=(18, 5), dpi=400)
# plt.plot(x, y)
# plt.plot(x, y, color="#748221")
# plt.plot(x, y, color="#748221", marker="o", markersize=40)
# plt.plot(x, y, color="#748221", marker="*", markersize=10, markerfacecolor="red")
plt.plot(x, y1, color="red", marker="*", markersize=10, mfc="blue", markeredgecolor="black", linestyle='dashed')
plt.plot(x, y2, color="#748221", marker="*", markersize=10, mfc="blue", markeredgecolor="black", linestyle='dashed')
# markeredgecolor="black" = mec="black"
plt.title("Graph Title", fontsize=20, color="blue")
plt.xlabel("X-axis", fontsize=15, color="green")
plt.ylabel("Y-axis", fontsize=15, color="green")
plt.xticks(x, ["2023", "2024", "2025", "2026"], fontsize=12, color="red")
plt.grid()
plt.show()
```

```python
plt.subplot(1, 2, 1)
plt.plot(x, y2)
plt.title("Graph 1")

plt.subplot(1, 2, 2)
plt.plot(x, y2)
plt.title("Graph 2")
```

```python
fig, axs = plt.subplots(2, 2, figsize=(18, 5))

axs[0, 0].plot(x, y1) # pr le graphe à activer entre les 4 graphes 
axs[0, 1].scatter(x, y1)
axs[1, 0].bar(x, y1)
fig.suptitle("Graph Title", fontsize=20, color="blue")
fig.savefig("graph.png")
```
